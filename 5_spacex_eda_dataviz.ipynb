# SpaceX Falcon 9 First Stage Landing Prediction
## Lab 5: Exploratory Data Analysis with Visualization

This notebook uses Matplotlib and Seaborn to see how flight number, payload mass, launch site and orbit relate to landing success, and then prepares the features for machine learning.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv")
print(df.shape)
df.head(5)
```

### Flight number vs payload mass
Each point is one launch. `Class = 1` means the first stage landed.

```python
sns.catplot(y="PayloadMass", x="FlightNumber", hue="Class", data=df, aspect=5)
plt.xlabel("Flight Number", fontsize=20)
plt.ylabel("Payload Mass (kg)", fontsize=20)
plt.title("Flight Number vs Payload Mass", fontsize=20)
plt.show()
```

### Task 1: Flight number vs launch site

```python
sns.catplot(y="LaunchSite", x="FlightNumber", hue="Class", data=df, aspect=5)
plt.xlabel("Flight Number", fontsize=20)
plt.ylabel("Launch Site", fontsize=20)
plt.title("Flight Number vs Launch Site", fontsize=20)
plt.show()
```

### Task 2: Payload mass vs launch site

```python
sns.catplot(y="LaunchSite", x="PayloadMass", hue="Class", data=df, aspect=5)
plt.xlabel("Payload Mass (kg)", fontsize=20)
plt.ylabel("Launch Site", fontsize=20)
plt.title("Payload Mass vs Launch Site", fontsize=20)
plt.show()
```

### Task 3: Success rate of each orbit type

```python
orbit_success = df.groupby('Orbit')['Class'].mean().reset_index().sort_values('Class', ascending=False)

plt.figure(figsize=(12, 6))
sns.barplot(x='Orbit', y='Class', data=orbit_success, color='steelblue')
plt.xlabel("Orbit", fontsize=14)
plt.ylabel("Success Rate", fontsize=14)
plt.title("Landing Success Rate by Orbit Type", fontsize=16)
plt.ylim(0, 1.05)
plt.show()

orbit_success
```

### Task 4: Flight number vs orbit type

```python
sns.catplot(y="Orbit", x="FlightNumber", hue="Class", data=df, aspect=5)
plt.xlabel("Flight Number", fontsize=20)
plt.ylabel("Orbit", fontsize=20)
plt.title("Flight Number vs Orbit Type", fontsize=20)
plt.show()
```

### Task 5: Payload mass vs orbit type

```python
sns.catplot(y="Orbit", x="PayloadMass", hue="Class", data=df, aspect=5)
plt.xlabel("Payload Mass (kg)", fontsize=20)
plt.ylabel("Orbit", fontsize=20)
plt.title("Payload Mass vs Orbit Type", fontsize=20)
plt.show()
```

### Task 6: Yearly trend of launch success

```python
df['Year'] = pd.to_datetime(df['Date']).dt.year
yearly_success = df.groupby('Year')['Class'].mean().reset_index()

plt.figure(figsize=(12, 6))
sns.lineplot(x='Year', y='Class', data=yearly_success, marker='o')
plt.xticks(yearly_success['Year'])
plt.xlabel("Year", fontsize=14)
plt.ylabel("Success Rate", fontsize=14)
plt.title("Launch Success Yearly Trend", fontsize=16)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()

yearly_success
```

### Features engineering
Select the features that will be used to predict landing success.

```python
features = df[['FlightNumber', 'PayloadMass', 'Orbit', 'LaunchSite', 'Flights', 'GridFins',
               'Reused', 'Legs', 'LandingPad', 'Block', 'ReusedCount', 'Serial']]
features.head()
```

### Task 7: Create dummy variables for the categorical columns

```python
features_one_hot = pd.get_dummies(features, columns=['Orbit', 'LaunchSite', 'LandingPad', 'Serial'])
print(features_one_hot.shape)
features_one_hot.head()
```

### Task 8: Cast all numeric columns to float64

```python
features_one_hot = features_one_hot.astype('float64')
features_one_hot.dtypes.value_counts()
```

```python
features_one_hot.to_csv('dataset_part_3.csv', index=False)
print("Saved dataset_part_3.csv with", features_one_hot.shape[0], "rows and", features_one_hot.shape[1], "columns")
```
